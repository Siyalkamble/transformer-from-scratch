    Tokenizer — start character-level (dead simple, no external deps, lets you debug shapes fast). Don't touch BPE/tiktoken yet — that's a distraction at this stage.
    
    Embedding + positional encoding — token embedding table + RoPE applied at attention time (not added to embeddings like sinusoidal — that's the mistake people carry over from old tutorials).
    
    Single attention head — Q, K, V projections, scaled dot product, causal mask (upper-triangular -inf mask before softmax). Test this in isolation with a tiny tensor and print shapes at every step. If you can't explain why the mask is applied before softmax and not after, stop and re-derive it.
    
    Multi-head attention — split into heads, run head 3 in parallel, concat, project through W_O.
    
    Transformer block — attention + residual + LayerNorm (pre-norm, not post-norm — pre-norm is what every modern LLM uses, post-norm is textbook-only and trains worse) + FFN + residual + LayerNorm + dropout. This is where your "dropout gap" from earlier applies — put it after attention output and after FFN, not randomly.
    
    Stack N blocks — 4-6 blocks is enough for a portfolio-scale model on your hardware (8GB RAM, CPU-only PyTorch — don't try to train something GPT-2-sized, you don't have the compute and nobody expects it).
    
    Training loop — cross-entropy loss on next-token prediction, AdamW, cosine LR schedule with warmup (this last part is commonly skipped by beginners and instantly signals "didn't read real training code" to anyone reviewing your repo).
    
    Sampling/generation — greedy first to verify correctness, then temperature + top-k/top-p once loss is sane.
